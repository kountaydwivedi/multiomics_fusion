```python

'''
plot confusion matrices

type the values by rounding off the average from the final curated excel sheets of each model/approach
'''


import seaborn as sns
import matplotlib.pyplot as plt

# Your confusion matrix
model = 'tabnet'
data = 'cnv'
cm = [[436, 47],
      [119, 386]]  # Replace this with your actual confusion matrix

# Optional: class labels
class_names = ['SCC', 'ADC']

# Set aesthetic style
sns.set(style="whitegrid")

# Create figure
plt.figure(figsize=(6, 5))
ax = sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", cbar=False,
                 annot_kws={"size": 25}, linewidths=1, linecolor='gray',
                 xticklabels=class_names, yticklabels=class_names)

plt.xlabel("Predicted", fontsize=20, labelpad=8)
plt.ylabel("Actual", fontsize=20, labelpad=8)
plt.title(f"Confusion_Matrix_{model}", fontsize=20, pad=8)
ax.tick_params(axis='both', labelsize=16)

plt.tight_layout()

# Save plots
plt.savefig(f"cm_{model}_{data}.png", dpi=600)
plt.savefig(f"cm_{model}_{data}.pdf", dpi=600)

plt.show()
```
